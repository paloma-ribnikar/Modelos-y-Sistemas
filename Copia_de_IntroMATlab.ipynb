Condiciones para nombrar variables
---
• El nombre debe comenzar con una letra del alfabeto. Después de eso, el nombre puede contener letras, dígitos y el carácter de subrayado (por ejemplo, valor_l), pero no puede tener un espacio.

• Existe un límite para la longitud del nombre; la función incorporada namelengthmax indica cuál es esta longitud máxima (los caracteres adicionales se truncan).

• MATLAB distingue entre mayúsculas y minúsculas. Entonces, las variables llamadas mynum, MYNUM y Mynum son todas diferentes (aunque esto sería confuso y no debería hacerse).

• Aunque los caracteres de subrayado son válidos en un nombre, su uso puede causar problemas con algunos programas que interactúan con MATLAB, por lo que algunos programadores usan mayúsculas y minúsculas en su lugar (por ejemplo, partWeights en lugar de part_weights).

• Hay ciertas palabras llamadas palabras reservadas, o palabras clave, que no se pueden usar como nombres de variables.

• Los nombres de las funciones integradas pueden, pero no deben, usarse como nombres de variables.

# RANDOM
Los generadores o funciones de números aleatorios no son verdaderamente aleatorios. Básicamente, la forma en que funciona es que el proceso comienza con un número, que se llama semilla. Con frecuencia, la semilla inicial es un valor predeterminado o se obtiene del reloj incorporado en la computadora. Luego, basándose en esta semilla, un proceso determina el siguiente "número aleatorio". Usando ese número como semilla la próxima vez, se genera otro número aleatorio y así sucesivamente. En realidad, se denominan **pseudoaleatorio**; no son realmente aleatorios porque hay un proceso que determina el siguiente valor cada vez.
La función **rand** se puede utilizar para generar números reales aleatorios distribuidos uniformemente; llamarlo genera un número real aleatorio en el intervalo abierto (0,1), lo que significa que los puntos finales del rango no están incluidos. No se pasan argumentos a la función rand en su forma más simple.

La semilla para la función rand siempre será la misma cada vez que se inicie MATLAB, a menos que se cambie la semilla inicial. La función **rng** establece la semilla inicial.
rng('shuffle')
con 'shuffle', la función rng usa la fecha y hora actuales para establecer la semilla, por lo que la semilla siempre será diferente. También se puede pasar un número entero para que sea la semilla.

Para generar random en un intervalo

rand* (high- low) + low

Para generar un entero (entre 0 y 10)

round (rand*10)

rand(n), crea una matriz de nxn random

randi genera enteros random



# OPERADOR :
Si los valores en el vector están espaciados regularmente, el operador de dos puntos puede usarse para iterar a través de estos valores.

Con el operador de dos puntos, también se puede especificar un valor de paso utilizando otros dos puntos, en la forma (primero: paso: último).

La función **linspace** crea un vector espaciado linealmente; linspace (x, y, n) crea un vector con n valores en el rango inclusivo de x a y. Si se omite n, el valor predeterminado es 100 puntos.

La función de ceros **zeros** crea una matriz de todos los ceros y la función **ones** una matriz de todos unos.


Para referirse a los elementos de la matriz, los subíndices de fila y luego de columna se dan entre paréntesis (siempre la fila primero y luego la columna). Puede usarse como indice **end** que indica el final de la fila o col.

toda la primera fila:
>> mat (1, :)

toda la segunda columna
mat(:,2)

mat= [a b]' traspone la matriz



La función de **length** devuelve el número de elementos de un vector. La función **size** devuelve el número de filas y columnas en un vector o matriz. **numel** el numero de elementos

La función **fliplr** "gira" la matriz de izquierda a derecha (en otras palabras, la columna más a la izquierda, la primera columna, se convierte en la última columna y así sucesivamente), y la función **flipud** cambia de arriba hacia abajo.

**La** función **abs** devolverá automáticamente el valor absoluto de cada elemento individual y el resultado será un vector con la misma longitud que el vector de entrada.

La función **min** devolverá el valor mínimo de un vector y la función **max** devolverá el valor máximo.

funciones **any** y **all**. La función any devuelve verdadero lógico si cualquier elemento en un vector representa verdadero, y falso si no. La función all devuelve verdadero lógico solo si todos los elementos representan verdadero.

La función **find** devuelve los índices de un vector que cumplen los criterios dados.

La función **isequal** es útil para comparar matrices. En MATLAB, el uso del operador de igualdad con matrices devolverá 1 o 0 para cada elemento.

La función **meshgrid** puede especificar las coordenadas xey de puntos en imágenes, o puede usarse para calcular funciones en dos variables xey.
Recibe como argumentos de entrada dos vectores y devuelve como argumentos de salida dos matrices que especifican por separado los valores xey.

# DEFINICION DE FUNCIONES
Al igual que los scripts, los archivos de funciones tienen la extensión .m. Aunque para ingresar definiciones de función en el Editor es posible elegir la flecha hacia abajo Nuevo y luego Función, será más fácil por ahora ingresar la función eligiendo Nuevo Script (esto ignora los valores predeterminados que se proporcionan cuando elige Función).
Una función en MATLAB consta de lo siguiente.

• El encabezado de la función (la primera línea), compuesto por:
la función de palabra reservada
el nombre del argumento de salida seguido del operador de asignación (=),
el nombre de la función (importante: debe ser el mismo
como el nombre del archivo en el que se almacena esta función para evitar confusiones)
los argumentos de entrada entre paréntesis, que corresponden a los argumentos que se pasan a la función en la llamada a la misma
•Un comentario que describe lo que hace la función (se imprime cuando se usa la ayuda)

•El cuerpo de la función, que incluye todas las declaraciones y, finalmente, debe poner un valor en el argumento de salida.

•terminar la función con **end** (tenga en cuenta que esto no es necesario en muchos casos en las versiones actuales de MATLAB, pero de todos modos se considera un buen estilo)

# ODE SOLVERS

ode45 is based on an explicit Runge-Kutta (4,5) formula, the Dormand-Prince pair. It is a one-step solver – in computing y(tn), it needs only the solution at the immediately preceding time point, y(tn-1). In general, ode45 is the best function to apply as a first try for most problems. [3]

ode23 is an implementation of an explicit Runge-Kutta (2,3) pair of Bogacki and Shampine. It may be more efficient than ode45 at crude tolerances and in the presence of moderate stiffness. Like ode45, ode23 is a one-step solver. [2]

ode113 is a variable order Adams-Bashforth-Moulton PECE solver. It may be more efficient than ode45 at stringent tolerances and when the ODE file function is particularly expensive to evaluate. ode113 is a multistep solver — it normally needs the solutions at several preceding time points to compute the current solution. [7]

The above algorithms are intended to solve nonstiff systems. If they appear to be unduly slow, try using one of the stiff solvers below.

ode15s is a variable order solver based on the numerical differentiation formulas (NDFs). Optionally, it uses the backward differentiation formulas (BDFs, also known as Gear's method) that are usually less efficient. Like ode113, ode15s is a multistep solver. Try ode15s when ode45 fails, or is very inefficient, and you suspect that the problem is stiff, or when solving a differential-algebraic problem. [9], [10]

ode23s is based on a modified Rosenbrock formula of order 2. Because it is a one-step solver, it may be more efficient than ode15s at crude tolerances. It can solve some kinds of stiff problems for which ode15s is not effective. [9]

ode23t is an implementation of the trapezoidal rule using a "free" interpolant. Use this solver if the problem is only moderately stiff and you need a solution without numerical damping. ode23t can solve DAEs. [10]

ode23tb is an implementation of TR-BDF2, an implicit Runge-Kutta formula with a first stage that is a trapezoidal rule step and a second stage that is a backward differentiation formula of order two. By construction, the same iteration matrix is used in evaluating both stages. Like ode23s, this solver may be more efficient than ode15s at crude tolerances. [8], [1]
